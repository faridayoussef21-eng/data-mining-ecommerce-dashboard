```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Pc\Desktop\e_commerce.csv")
print(df.head())
df = df.drop_duplicates()
df['CustomerNo'] = df['CustomerNo'].fillna(0).astype(int)
df['Date'] = pd.to_datetime(df['Date'])
df = df[df['Quantity'] > 0]
df['ProductName'] = df['ProductName'].str.strip()
df['Country'] = df['Country'].str.strip()

print(df.info())
df.to_csv(r"C:\Users\Pc\Desktop\cleaned_e_commerce.csv", index=False)
```
